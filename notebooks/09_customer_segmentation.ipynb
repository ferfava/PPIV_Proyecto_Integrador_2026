Objetivo: descubrir perfiles de clientes sin usar `churn` para crear los grupos. Luego se analiza la tasa de churn de cada segmento solo con fines de interpretación.

```python
from pathlib import Path
import sys
import pandas as pd

ROOT = Path('..').resolve()
sys.path.append(str(ROOT))
from scripts.run_customer_segmentation import run_segmentation
```

```python
df = pd.read_csv(ROOT / 'data' / 'raw' / 'sales_marketing_customer_raw.csv')
result = run_segmentation(df)
result['comparison'].round(4)
```

## Selección de número de clusters

Se comparan valores de k entre 2 y 8 usando silhouette, Davies-Bouldin, Calinski-Harabasz e inercia. Se selecciona k=4 por equilibrio entre calidad interna e interpretabilidad de negocio.

```python
result['stability'].round(4)
```

## Perfil de segmentos

```python
profile = result['profile'].copy()
profile['churn_rate'] = (profile['churn_rate'] * 100).round(2)
profile['share'] = (profile['share'] * 100).round(2)
profile.round(2)
```
