# 07 - Modelos avanzados y ensemble

Comparación de modelos avanzados para predicción de churn con validación cruzada estratificada, búsqueda de hiperparámetros y ensemble de probabilidades.

```python
from pathlib import Path
import sys
import pandas as pd

ROOT = Path('..').resolve()
sys.path.append(str(ROOT))

from scripts.run_advanced_models import run_advanced_models
```

```python
DATA_PATH = ROOT / 'data' / 'raw' / 'sales_marketing_customer_raw.csv'
df = pd.read_csv(DATA_PATH)
result = run_advanced_models(df)
```

## 1. Comparación mediante validación cruzada

```python
display(result['cv_results'][['model','accuracy','balanced_accuracy','precision','recall','f1','roc_auc','pr_auc']].round(4))
```

## 2. Optimización de XGBoost

Se usa una búsqueda aleatoria acotada y reproducible. La selección se realiza por PR-AUC y luego ROC-AUC, métricas más informativas que accuracy ante el desbalance.

```python
print(result['best_xgb_params'])
display(result['tuning'].head(5).round(4))
```

## 3. Evaluación final sobre test

```python
cols=['model','accuracy','balanced_accuracy','precision','recall','f1','roc_auc','pr_auc','tn','fp','fn','tp']
display(result['test_results'][cols].round(4))
```

## 4. Ensemble de probabilidades

Se promedian las probabilidades de Gradient Boosting, XGBoost y CatBoost. El objetivo es reducir la dependencia de un único algoritmo y capturar señales complementarias. El ensemble se conserva únicamente si mejora la calidad predictiva.

## Conclusión

Los modelos basados en boosting superan claramente al baseline logístico. El ensemble ofrece el mejor equilibrio global en test y será candidato principal para las siguientes etapas de calibración, explicabilidad y decisión de negocio.
