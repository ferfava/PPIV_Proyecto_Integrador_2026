## Decisión de calibración

Se conserva **Platt scaling** porque reduce de forma marcada Brier Score y ECE sin alterar el orden de los scores, por lo que mantiene ROC-AUC y PR-AUC. La calibración isotónica es ligeramente mejor en calibración pura, pero introduce pequeñas pérdidas de discriminación por empates y es menos suave.

## Interpretación de negocio

El análisis muestra que un umbral fijo de 0,50 no es necesariamente la política adecuada. En escenarios donde perder un cliente cuesta más que contactar a uno que finalmente no abandona, conviene una política más agresiva. Para operaciones con capacidad limitada, es más estable priorizar por **ranking** (Top 10%, 20%, 25%, etc.) que elegir un umbral universal.
