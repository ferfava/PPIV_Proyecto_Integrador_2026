# 06 - Baseline Models

Se construyen modelos base para establecer un piso de comparación antes de entrenar algoritmos más complejos.

La clase `churn=1` representa aproximadamente 15 % de los casos, por lo que **accuracy no será la métrica principal**.


```python
from pathlib import Path
import sys
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path('..')
sys.path.append(str(ROOT.resolve()))

from scripts.run_baseline_models import run_baselines

DATA_PATH = ROOT / 'data' / 'raw' / 'sales_marketing_customer_raw.csv'
df = pd.read_csv(DATA_PATH)
```

## 1. Entrenamiento y evaluación


```python
prepared, metrics, coefficients = run_baselines(df)
metrics.round(4)
```

### Lectura

El Dummy puede obtener una accuracy alta simplemente prediciendo siempre la clase mayoritaria. Por eso se comparan también balanced accuracy, recall, F1, ROC-AUC y PR-AUC.


## 2. Matrices de confusión resumidas


```python
metrics[['model', 'tn', 'fp', 'fn', 'tp']].set_index('model')
```

## 3. Comparación de métricas


```python
plot_cols = ['balanced_accuracy', 'precision', 'recall', 'f1', 'roc_auc', 'pr_auc']
ax = metrics.set_index('model')[plot_cols].T.plot(kind='bar', figsize=(12, 6))
ax.set_ylim(0, 1)
ax.set_ylabel('Score')
ax.set_title('Comparación de modelos baseline')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 4. Interpretabilidad de la Regresión Logística balanceada


```python
coefficients.head(15)[['feature', 'coefficient']].reset_index(drop=True)
```

Los coeficientes se interpretan en el contexto del preprocessing aplicado. Las variables numéricas fueron estandarizadas y las categóricas codificadas con one-hot encoding.

Las señales más fuertes son coherentes con el EDA: gasto y satisfacción reducen el riesgo estimado, mientras que más tickets de soporte lo incrementan. Esto no implica causalidad.


## Conclusión

La Regresión Logística balanceada constituye un baseline serio porque:

- supera ampliamente al Dummy en detección de churn;
- recupera una proporción alta de la clase positiva;
- conserva interpretabilidad;
- proporciona probabilidades que luego podrán calibrarse;
- establece un punto de referencia para modelos avanzados.
