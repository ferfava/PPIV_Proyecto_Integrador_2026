# 10 - Retention Engine

Este notebook convierte las predicciones de churn en un ranking operativo de clientes, combinando riesgo, valor comercial y segmento.

```python
from pathlib import Path
import sys
import pandas as pd

ROOT = Path('..').resolve()
sys.path.append(str(ROOT))
from scripts.run_retention_engine import build_retention_engine

df = pd.read_csv(ROOT / 'data' / 'raw' / 'sales_marketing_customer_raw.csv')
result = build_retention_engine(df)
```

## 1. Métricas del modelo de riesgo

```python
pd.Series(result['model_metrics']).round(4)
```

## 2. Evaluación del ranking

El objetivo no es solo clasificar, sino concentrar la mayor cantidad posible de churn reales en una lista corta de clientes priorizados.

```python
result['evaluation'].round(4)
```

## 3. Distribución por bandas de prioridad

```python
result['ranking']['priority_band'].value_counts().sort_index()
```

## 4. Clientes con mayor prioridad

```python
cols = ['customer_id', 'churn_probability', 'value_score', 'segment', 'priority_score', 'priority_band', 'recommended_action']
result['ranking'][cols].head(20).round(4)
```
