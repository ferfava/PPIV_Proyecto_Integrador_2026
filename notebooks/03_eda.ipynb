## Principales hallazgos

1. La tasa global de churn es 15,32 %.
2. Satisfacción 1–2 concentra aproximadamente 50 % de churn, frente a cerca de 9 % en satisfacción 3–5.
3. Los clientes con 5+ tickets de soporte presentan aproximadamente 50 % de churn.
4. El quintil de menor gasto presenta aproximadamente 40,5 % de churn, muy por encima del resto.

## Hipótesis candidatas

- **H1:** satisfacción baja (1–2) se asocia con mayor churn que satisfacción 3–5.
- **H2:** 5 o más tickets de soporte se asocian con mayor churn que 0–4 tickets.
- **H3:** pertenecer al quintil inferior de `total_spent` se asocia con mayor churn que pertenecer a los cuatro quintiles superiores.

Estas asociaciones son descriptivas y no implican causalidad. Se contrastarán estadísticamente en `04_hypothesis_testing.ipynb`.
